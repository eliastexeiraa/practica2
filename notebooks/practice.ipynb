```python
import matplotlib.pyplot as plt
import numpy as np

# Tu código a partir de aquí
```

```python
normal = np.random.normal(loc=0, scale=1, size=1000)
uniforme = np.random.uniform(low=-3, high=3, size=1000)
```

```python
print("Distribución normal")
print(f"Media: {np.mean(normal):.3f}")
print(f"Mediana: {np.median(normal):.3f}")
print(f"Varianza: {np.var(normal):.3f}")
print(f"Desvío estándar: {np.std(normal):.3f}")
print(f"Mínimo: {np.min(normal):.3f}")
print(f"Máximo: {np.max(normal):.3f}")
```

```python
print("Distribución uniforme")
print(f"Media: {np.mean(uniforme):.3f}")
print(f"Mediana: {np.median(uniforme):.3f}")
print(f"Varianza: {np.var(uniforme):.3f}")
print(f"Desvío estándar: {np.std(uniforme):.3f}")
print(f"Mínimo: {np.min(uniforme):.3f}")
print(f"Máximo: {np.max(uniforme):.3f}")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].hist(normal, bins=30, color="steelblue", alpha=0.8, label="Normal")
axes[0].set_title("Distribución normal")
axes[0].set_xlabel("Valor")
axes[0].set_ylabel("Frecuencia")
axes[0].legend()
axes[0].grid(True, alpha=0.3)

axes[1].hist(uniforme, bins=30, color="darkorange", alpha=0.8, label="Uniforme")
axes[1].set_title("Distribución uniforme")
axes[1].set_xlabel("Valor")
axes[1].set_ylabel("Frecuencia")
axes[1].legend()
axes[1].grid(True, alpha=0.3)

fig.suptitle("Comparación de distribuciones aleatorias")
plt.tight_layout()
plt.show()
```

La distribución normal se concentra alrededor de la media, cercana a 0, y presenta una forma de campana. Sin embargo, la distribución uniforme reparte los valores de manera más pareja entre -3 y 3. La uniforme muestra mayor dispersión, lo que también se observa en su varianza, cercana a 3, frente a una varianza cercana a 1 en la normal.
